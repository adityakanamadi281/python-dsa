# What is the difference between List and Tuples ?

## Lists :
* Lists are represented in square bracket,
* Lists are mutable
* List is less efficient in memory utilization than tuples
* Lists are slow in execution than tuples
* Comprehension is only applicable for lists
* List is supports for packaging but not  for unpackaging



## Tuples :
* Tuples are represented in paranthesis
* Tuples are immutable
* Tuple objects takes less memory than list objects for same data
* Comprehension is not applicable for tuples
* Tuples are supports for both packaging and unpackaging.

```python
# write a code to display the words that are repeated more than or equal to N times in the text..


def count_words():
  str1 = input("Enter your words :  ").split()
  n=int(input("Enter the N times number : "))

  counts = dict()

  for word in str1:
    if word in counts:
      counts[word] += 1
    else:
      counts[word]=1

  word_list=[]

  for key in counts:
    if counts[key]>=n:
      word_list.append(key)

  if len(word_list)>0:
    print(f"Words with {n} Times : {word_list}")
  else:
    print("NA")

count_words()
```

```python
# write a code to to sort charecters  and numbers so that first alphabets and then numbers are  printed

# input = "A7B1R3"
# output = "ABR137"


input = "A7B1R3"

alphabets = []
numbers = []

for ch in input:
  if ch.isalpha():
    alphabets.append(ch)
  else:
    numbers.append(ch)

final_list = sorted(alphabets) + sorted(numbers)

output = "".join(final_list)

print(output)
```

```python
# write a code to print armstrong numbers between 1 to 100

for i in range(1,1001):
  num=i
  n=len(str(i))
  sum1=0

  i=str(i)

  for digit in i:
    sum1 = sum1 + int(digit)**n

  if sum1==num:
    print(num)
```
