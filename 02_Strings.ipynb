```python
# reverse a string

def rev_string():
  s=str(input("Enter a string : ")).lower()

  rev_string=""

  for ch in s:
    rev_string = ch + rev_string

  return rev_string

rev_string()
```

```python
# check if a string is palindrome

def check_str_palindrome():
    s = str(input("Enter  a string : "))

    rev_str = s[::-1]

    if s==rev_str:
      print(f"{s} is a palindrome")
    else:
      print(f"{s} is not palindrome")


check_str_palindrome()
```

```python
# count the frequency of charecters in a string

def count_freq_charecters():
    s = str(input("Enter a string : "))
    freq = {}

    for ch in s:
      if ch.isalpha():
        freq[ch]=freq.get(ch, 0) + 1

    return freq


count_freq_charecters()
```

```python
# find the first non-repeating charecter in a string

def non_repeating_ch():
    s = str(input("Enter a string : "))
    freq = {}

    for ch in s:
      freq[ch]=freq.get(ch,0) + 1

    for ch in s:
      if freq[ch]==1:
        return ch
    return None


non_repeating_ch()
```

```python
# find second non-repeating charecter

# find the first non-repeating charecter in a string

def non_repeating_ch():
    s = str(input("Enter a string : "))
    freq = {}

    count = 0

    for ch in s:
      freq[ch]=freq.get(ch,0) + 1

    for ch in s:
      if freq[ch]==1:
        count += 1
        if count==2:
          return ch
    return None


non_repeating_ch()
```

```python
# check the the given charecter is vowel or consonant

def check_ch():
    ch=str(input("Enter charecter : "))

    vowels="aeiouAEIOU"

    if ch in vowels:
      return "Vowel"
    elif ch.isalpha():
      return "Consonant"
    else:
      return "Invalid"

check_ch()
```

```python
# Count the vowels and consonants in a sentence

def count_vowel_consonant():
    s=str(input("Enter the  sentence : ")).lower()
    vowels = "aeiouAEIOU"
    vowel_count = 0
    consonant_count = 0

    for ch in s:
      if ch in vowels:
        vowel_count += 1
      elif ch.isalpha():
        consonant_count += 1

    return vowel_count, consonant_count

count_vowel_consonant()
```

```python
# nth fibonacci number

def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)

fib(6)
```
