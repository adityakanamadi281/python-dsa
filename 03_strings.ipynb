```python

# Reverse the charecters in string

def rev_str():
    string = str(input("Enter a string : "))
    reversed_string = " "

    for ch in string:
      reversed_string = ch + reversed_string

    return reversed_string

rev_str()
```

```python
# Reverse the words in a sentence without reversing individual characters

def rev_words():
    sentence = str(input("Enter a sentence : "))

    words = sentence.split()
    reversed_sentence = " ".join(words[::-1])
    return reversed_sentence

rev_words()
```

```python
# check two strings are anagrams

def check_anagrams():
  str1 = str(input("Enter string 1 : ")).lower()
  str2 = str(input("Enter string 2 : ")).lower()

  if sorted(str1)==sorted(str2):
    print("Strings are anagrams")
  else:
    print("Strings are not anagrams ")

check_anagrams()

```

```python
# find the most frequent character

def most_freq_ch():
    string = input("Enter the string: ").lower()

    freq = {}

    for ch in string:
        freq[ch] = freq.get(ch, 0) + 1

    most_ch = max(freq, key=freq.get)

    print(f"The most frequent character is '{most_ch}'")
    print(f"The frequency of the character is {freq[most_ch]}")

most_freq_ch()
```

```python
# Remove all whitespaces from a string

def remove_ws():
  string = str(input("Enter a string : "))

  return  "".join(string.split())

remove_ws()
```
