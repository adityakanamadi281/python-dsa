```python
# Find the maximum elemennt in a list

def max_element():
    lst = input("Enter your list : ")

    maximum = lst[0]

    for i in lst:
      if i>maximum:
        maximum = i
    print(f"Maximum element : {maximum}")


max_element()
```

```python
# Reverse a list in place

def rev_list():
    lst = input("Enter list elements : ")

    rev_list = lst[::-1]
    return rev_list

rev_list()
```

```python
# Finnd the sum and average of a list

def sum_and_avg():
    lst = input("Enter a list elements : ").split()

    total = 0

    for i in lst:
      total += int(i)

    avg = total / len(lst)

    return total, avg

total, avg = sum_and_avg()

print(f"Sum of list : {total}")
print(f"Average of list : {avg}")
```
